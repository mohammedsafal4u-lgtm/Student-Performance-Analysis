<h2 style="color:green; font-weight:bold;">✏️INTRODUCTION</h2>


**Student academic performance is a key indicator of the effectiveness of an education system and the overall development of learners. Educational institutions continuously aim to improve student outcomes; however, academic success is influenced by multiple interrelated factors rather than classroom teaching alone. Factors such as study habits, availability of learning resources, family background, time management, and social behavior play an important role in shaping students’ academic achievements.**

**The dataset used in this project contains detailed information related to students’ academic, personal, and social attributes. It includes variables such as study hours, access to transportation and digital devices, parental education status, use of smartphones and personal computers, time spent on social media, attendance behavior, and disciplinary records such as probation or suspension history. These attributes provide a comprehensive view of the conditions that may influence student learning and performance.**

**The purpose of this project is to analyze the given dataset to understand how these factors impact student academic performance. By applying data preprocessing techniques, exploratory data analysis, and visualization methods, meaningful patterns and relationships are identified within the data. This helps in recognizing both positive and negative influences on student outcomes.**

**The dataset used in this study, titled “Students Performance Data Set,” contains 1195 individual student responses with 30 features captured through an online survey. Each row represents the academic and non-academic profile of a student from a private university in Bangladesh, specifically from the Computer Science and Engineering (CSE) department. In total, the dataset holds 37,045 data points, offering a rich source of information for analysis and this data set taken from "https://data.mendeley.com/"**

## **⭐STEP 7.MULTIVARIATE ANALYSIS**

**Multivariate Analysis** is a part of Exploratory Data Analysis (EDA) that involves analyzing three or more variables simultaneously. The main objective of multivariate analysis is to understand complex relationships, interactions, and combined effects of multiple variables on the dataset.

**📊7.1 Pair Plot (CGPA, Study Hours, Age, Gender)**

```python
sns.pairplot(
    df[['Current_CGPA',
        'Daily_Skill_Development_Hours',
        'Age',
        'Gender']],
    hue='Gender'
)
plt.show()
```

**Interpretation of the Pair Plot:**

🎯The scatter plots reveal a positive relationship between daily study hours and CGPA, where higher study time generally corresponds to higher academic performance.

🎯The relationship between age and CGPA appears weak, suggesting that age does not significantly influence academic outcomes.

🎯Points are colored by gender, and the overlapping patterns indicate that both genders follow similar trends in study habits and academic performance.

🎯No clear separation between genders is observed, reinforcing that gender is not a strong differentiating factor in CGPA.

**Findings:** The pair plot highlights a positive association between study hours and CGPA, minimal influence of age on academic performance, and similar performance patterns across genders.

**📊7.2 Study Level vs CGPA by Gender**

**📊7.3 CGPA Category by Gender – Stacked Bar Chart**

```python
cgpa_gender = pd.crosstab(
    df['CGPA_Category'],
    df['Gender']
)

cgpa_gender.plot(kind='bar', stacked=True)
plt.title("CGPA Category Distribution by Gender")
plt.xlabel("CGPA Category")
plt.ylabel("Number of Students")
plt.show()
```

**Findings:The stacked bar chart shows a similar distribution of CGPA categories across genders, with most students falling under good and excellent performance levels, indicating balanced academic achievement.**
